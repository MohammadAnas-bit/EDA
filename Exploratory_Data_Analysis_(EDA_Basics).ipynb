**Q1. What is the average, median, and standard deviation of house prices?**

```python
# Average Price

df['Price'].mean()
```

```python
# Median Price

df['Price'].median()
```

```python
# Standard Deviation

df['Price'].std()
```

**Q2. Which number of bedrooms is most common?**

```python
print ('No. of bedrooms most common')
df['number of bedrooms'].mode()[0]
```

**Q3. Check if the price data is skewed.**

```python
skewness= df['Price'].skew()
print ('Skewness:',skewness)
```

```python
if skewness >0 :
  print ('Price data is positively skewed')
elif skewness <0:
  print ('Price data is negatively skewed')
else :
  print ('Price data is normally distributed')
```

**Q4. What is the average price for each number of bedrooms?**

```python
df.groupby('number of bedrooms')['Price'].mean().round(2)
```

**Q5.  What is the relationship between living area and price? (Use : Area without Basement).**

```python
df[['Area of the house(excluding basement)','Price']].corr()
```

**Q6. Identify any anomalies where houses have high prices but low areas.**

```python
df[['Area of the house(excluding basement)','Price']].sort_values(['Price','Area of the house(excluding basement)'], ascending=[False,True]).head(10)
```

These houses have relatively high prices despite having comparatively smaller areas, so they can be considered potential price area anomalies.


**Q7. Compare average price based on number of floors and houses with or without waterfront.**

```python
print('Average price by number of floors and waterfront presence:')
display(df.groupby(['number of floors' , 'waterfront present'])['Price'].mean().unstack().round(2))
```

**Q8. Identify the minimum and maximum house price. What does this indicate?**

```python
df['Price'].min()
```

```python
df['Price'].max()
```

The minimum house price of `78000` and the maximum house price of `7700000` indicate a very wide range in property values within this dataset. This suggests a diverse housing market, likely including properties from various segments, such as small, affordable homes to very large, luxurious estates. This broad price range is important to consider when performing further analysis, as it implies significant variation in factors influencing house prices.

**Q9. Which location (zipcode/area) has the highest average price?**

```python
df.groupby('Postal Code')['Price'].mean().round(2).sort_values(ascending=False).head(1)
```
